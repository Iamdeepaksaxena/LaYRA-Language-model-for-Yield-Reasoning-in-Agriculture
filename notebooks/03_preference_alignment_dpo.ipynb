### Comparing Responses

```python
def compare_models(question):
    
    generation_args = {
        "max_new_tokens": 100,
        "temperature": 0.8,
        "top_p": 0.9,
        "do_sample": True,
        "repetition_penalty": 1.1,
    }

    print("=" * 100)
    print("QUESTION:")
    print(question)
    print("=" * 100)

    models = {
        "Non-Instruction": non_instruction_model,
        "Instruction": instruction_model,
        "Preference-Aligned": preference_model,
    }

    for name, model in models.items():

        inputs = tokenizer(question, return_tensors="pt").to(model.device)

        with torch.no_grad():
            outputs = model.generate(
                **inputs,
                **generation_args
            )

        response = tokenizer.decode(
            outputs[0],
            skip_special_tokens=True
        )

        print(f"\n{name} Model Output:\n")
        print(response)
        print("-" * 100)
```

#### function for response generation from all models

```python
import torch


def generate_response(model, question):

    model.eval()

    # get model device
    device = next(model.parameters()).device


    inputs = tokenizer(
        question,
        return_tensors="pt"
    ).to(device)


    with torch.no_grad():

        outputs = model.generate(
            **inputs,
            max_new_tokens=150,
            do_sample=False,
            repetition_penalty=1.1
        )


    response = tokenizer.decode(
        outputs[0],
        skip_special_tokens=True
    )


    if question in response:
        response = response.replace(question, "")


    return response.strip()
```
